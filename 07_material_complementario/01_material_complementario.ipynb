### Ejercicio 1: Identificación y Registro de Vehículos

**Objetivo:** Declarar variables utilizando cadenas de texto (`str`), números enteros (`int`) y flotantes (`float`), combinándolos mediante interpolación de texto (`f-strings`).

```python
# Declaración de variables del sistema de navegación
modelo_vehiculo = "Tesla Model Y"
anio_fabricacion = 2024
velocidad_actual = 85.5
motor_encendido = True

# Presentación de datos por consola utilizando f-strings
print(f"Vehículo registrado: {modelo_vehiculo} ({anio_fabricacion})")
print(f"Velocidad de crucero actual: {velocidad_actual} km/h")
print(f"Estado del motor: {motor_encendido}")
```

### Ejercicio 2: Validación de Estados Lógicos

**Objetivo:** Utilizar tipos de datos booleanos (`bool`) y operadores lógicos para evaluar las condiciones iniciales de seguridad antes de arrancar.

```python
# Variables de estado del sistema
puertas_cerradas = True
cinturon_conductor = True
freno_mano_activado = False

# Evaluamos condiciones lógicas simples
sistema_listo = puertas_cerradas and cinturon_conductor and not freno_mano_activado

print(f"¿Cumple con todas las normas de seguridad?: {sistema_listo}")
```
