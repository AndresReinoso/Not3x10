> ## **¿Cómo guarda Python la información en la memoria?** 

<br>


### Conceptos Clave:

<br>

1. **¿Qué es una variable?:**

Es un espacio reservado en la memoria del computador al que se le asigna un nombre identificador para almacenar un valor, permitiendo consultarlo o modificarlo durante la ejecución del programa.

2. **Los tipos de datos básicos en Python:**

   * **Texto (`str`):** Para nombres o letras (ej. `"Tesla Model 3"`). Siempre van entre comillas.
   * **Números con decimales (`float`):** Para mediciones precisas o continuas (ej. la velocidad `78.4` km/h).
   * **Números enteros (`int`):** Para conteos exactos (ej. número de puertas abiertas: `0`).
   * **Booleanos (`bool`):** Interruptores lógicos que solo aceptan dos estados: Verdadero (`True`) o Falso (`False`).
</n>

3. **El signo `=`:** En programación no significa "igualdad matemática", sino una orden: **"Toma lo que está a la derecha y guárdalo dentro de la caja de la izquierda"**.

> ## **Ejemplo 1** 

```python
# Registrando datos del panel de control de un automóvil en la memoria

modelo_auto = "Tesla Model 3"          # Texto (str)
velocidad_actual_kmh = 78.4            # Número decimal (float)
puertas_abiertas = 0                   # Número entero (int)
motor_encendido = True                 # Booleano / Interruptor (bool)
```

Como puedes observar, asignamos los valores usando el signo `=`. Arriba encontrarás los tipos de datos más sencillos de Python.

```python
# Comprobamos qué tipo de dato guardó Python en cada variable usando type()

print("Tipo de dato del modelo:", type(modelo_auto))
print("Tipo de dato de la velocidad:", type(velocidad_actual_kmh))
print("Tipo de dato del motor:", type(motor_encendido))
```

Puedes emplear el comando print, para mostrar valores en pantalla, de igual manera el separador `,`, permite suministrar varios valores, textos o resultados de funciones en una misma línea para que la computadora los imprima juntos. El comando type(), permite obtener el tipo de dato almacenado.

> ## **Ejemplo 2** 

Podemos imprimir usando un formato específico (**f-strings**), el cual nos permite insertar variables directamente dentro del texto utilizando llaves `{}`.

```python
# Mostramos el reporte limpio en la pantalla del vehículo

print("")
print("--- ESTADO DEL VEHÍCULO ---")
print("")
print(f"Modelo: {modelo_auto}")
print(f"Velocidad Actual: {velocidad_actual_kmh}kmh")
print(f"¿El motor está encendido?: {motor_encendido}")
```
